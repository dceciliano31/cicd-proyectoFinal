```python
# Databricks notebook source

# COMMAND ----------
dbutils.widgets.removeAll()

dbutils.widgets.text("catalog_name", "proyectoFinal")
dbutils.widgets.text("raw_schema", "raw")
dbutils.widgets.text("bronze_schema", "bronze")
dbutils.widgets.text("silver_schema", "silver")
dbutils.widgets.text("gold_schema", "gold")
dbutils.widgets.text("storage_account", "asaproyecto")

# COMMAND ----------
catalog_name = dbutils.widgets.get("catalog_name")
raw_schema = dbutils.widgets.get("raw_schema")
bronze_schema = dbutils.widgets.get("bronze_schema")
silver_schema = dbutils.widgets.get("silver_schema")
gold_schema = dbutils.widgets.get("gold_schema")
storage_account = dbutils.widgets.get("storage_account")

data_uri = f"abfss://raw@{storage_account}.dfs.core.windows.net/"

print("Configuración:")
print(f"catalog_name={catalog_name}")
print(f"raw_schema={raw_schema}")
print(f"bronze_schema={bronze_schema}")
print(f"silver_schema={silver_schema}")
print(f"gold_schema={gold_schema}")
print(f"storage_account={storage_account}")

# COMMAND ----------
sql_statements = [
    f"""
    CREATE CATALOG IF NOT EXISTS {catalog_name}
    COMMENT 'Catálogo del proyecto final de ingeniería de datos'
    """,
    f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{raw_schema}",
    f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{bronze_schema}",
    f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{silver_schema}",
    f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{gold_schema}",
]

for stmt in sql_statements:
    print("Ejecutando:")
    print(stmt.strip())
    spark.sql(stmt)

print("Catálogo y schemas creados correctamente.")
```
